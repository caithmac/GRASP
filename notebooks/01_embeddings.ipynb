# GRASP molecular embeddings

In Colab, run all cells in order; the first code cell installs GRASP. Locally, run from the repository root after `pip install -e .`. The pretrained encoder returns representations, not property predictions. This notebook uses local release weights when present; otherwise it downloads `caithmac/GRASP`.


```python
import os
import subprocess
import sys
from pathlib import Path

try:
    import google.colab
except ImportError:
    pass
else:
    repo = Path('/content/GRASP')
    if not repo.is_dir():
        subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/caithmac/GRASP.git', str(repo)], check=True)
    os.chdir(repo)
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', '.'], check=True)
```

```python
from pathlib import Path
from grasp import GRASPEncoder

model_source = 'hf_model' if Path('hf_model/model.safetensors').is_file() else 'caithmac/GRASP'
encoder = GRASPEncoder.from_pretrained(model_source)
smiles = ['CCO', 'CC(=O)O', 'c1ccccc1']
embeddings = encoder.encode(smiles)
print(embeddings.shape)  # (3, 768)
```

```python
import numpy as np

unit = embeddings / np.linalg.norm(embeddings, axis=1, keepdims=True)
print(np.round(unit @ unit.T, 3))  # illustrative cosine similarities
```
