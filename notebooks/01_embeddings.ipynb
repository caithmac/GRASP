# GRASP molecular embeddings

Run from the repository root after `pip install -e .`. The pretrained encoder returns representations, not property predictions. This notebook uses local release weights when present; otherwise it downloads `caithmac/GRASP`.


```python
from pathlib import Path
from grasp import GRASPEncoder

model_source = 'hf_model' if Path('hf_model/model.safetensors').is_file() else 'caithmac/GRASP'
encoder = GRASPEncoder.from_pretrained(model_source)
smiles = ['CCO', 'CC(=O)O', 'c1ccccc1']
embeddings = encoder.encode(smiles)
print(embeddings.shape)  # (3, 768)
```

```python
import numpy as np

unit = embeddings / np.linalg.norm(embeddings, axis=1, keepdims=True)
print(np.round(unit @ unit.T, 3))  # illustrative cosine similarities
```
